```python
import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
from sklearn.metrics import mean_absolute_error, mean_absolute_percentage_error,root_mean_squared_error
from sklearn.linear_model   import LinearRegression
from sklearn import linear_model, metrics
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("car_prices.csv")
```

```python
df.head()
```

```python
df.describe()
```

```python
df.info()
```

```python
sns.pairplot(df)
plt.suptitle("Pariplot of Features and target", y=1.02)            #title for whole pairplot
plt.show()
```

```python
# price = (m1 * engine_capacity) + (m2 * age) + (m3 * mileage) + b
```

```python
X = df[['engine_capacity', 'age', 'mileage']]
y=df['price']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
print(f"Traininsg samples: {X_train.shape[0]}")
```

```python
model=LinearRegression()
model.fit(X_train, y_train)
```

```python
dir(model)
```

```python
print("Intercept",model.intercept_)
print("Coeff: ",model.coef_)
```

```python
#prediction on test set
y_pred = model.predict(X_test)

#prdict for new car
sample_pred = model.predict([[900, 5, 12.5]])
print(f"Predicted price for 900 CC, 5years, 12.5km/l: {sample_pred[0]:.2f}")
```

```python
r2 = model.score(X_test, y_test)
print(f"R2 score: {r2:.3f}")
```

```python
mae = metrics.mean_absolute_error(y_test, y_pred)
mse = metrics.mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)

print(f"MAE: {mae: .2f}")
print(f"MSE: {mse:.2f}")
print(f"RMSE: {rmse:.2f}")
```

```python
plt.figure(figsize=(8,5))
plt.scatter(y_test, y_pred, color='blue', alpha=0.7)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--')
plt.xlabel('Actual price')
plt.ylabel('Predicted price')
plt.title('Actual vs Predicted price')
plt.show()
```

Exercise: Multiple Linear Regression

Problem Statement

A real estate company in India wants to predict the selling price of houses based on their characteristics. They have collected data for houses including:

* area (in square feet)
* bedrooms (number of bedrooms)
* age (age of the house in years)
* price (in lakhs INR)

The goal is to build a Multiple Linear Regression model to predict house prices.

```python
df2 = pd.read_csv("house_prices_mlr.csv")
```
