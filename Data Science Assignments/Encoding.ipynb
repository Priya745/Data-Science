```python
# Encode categorical variables into numerical form

#EDA Example
import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
import seaborn as sns

data = {'Transaction_ID': [1, 2, 3, 4, 5],
        'Quantity': [2, 1, 3, 4, 2],
        'Price_per_item': [10.50, 25.00, 5.75, 12.00, 15.50],
        'Customer_ID': ['C001', 'C002', 'C001', 'C003', 'C002']}

df = pd.DataFrame(data)

print(df)

```

```python
df['total_amount'] = df['Quantity']  * df['Price_per_item']
print(df)
```

```python
df['Transaction_Date'] = pd.to_datetime(['2026-03-04', '2025-06-12', '2024-12-23', '2020-09-30', '2021-04-09'])


df['Transaction_year'] = df['Transaction_Date'].dt.year
df['Trascation_month'] = df['Transaction_Date'].dt.month
# df['Transaction_day_of_week'] = df
print(df)
```

```python
df_encoded = pd.get_dummies(df,columns=['Customer_ID'], prefix='Customer', dtype=int)

#get_dummies() is a Pandas function that helps you convert categorical variables into numeric form by creating dummy/indicator variables.

print("DataFrame with one hot encoded 'Customer_ID:")
df_encoded
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
import warnings
warnings.filterwarnings('ignore')

df2 = pd.read_csv('/content/car_prices_ohe.csv')

df2
```

```python
print('Unique brands:', df2['brand'].unique())
```

```python
df2.describe()
```

```python
df_dummies=pd.get_dummies(df2, columns=['brand'], dtype=int)
df_dummies.head()
```

```python
#avodidng the dummy variable trap

df_dummies = df_dummies.drop('brand_Toyota', axis=1)
df_dummies
```

```python
# x is engine size and brand
# y is price
```

```python
X = df_dummies.drop('price', axis=1)
y = df_dummies['price']
print(X)
print(y)
# model = LinearRegression()
# model.fit(X,y)
```

```python
model = LinearRegression()
model.fit(X,y)
```

```python
print('Model coeffecients:', model.coef_)
print("Model intercept:", model.intercept_)
print('R2 score:', model.score(X,y))

# y = bx+c = b1x1 + b2x2 + b3x3 + c
```

```python
pred = model.predict([[2.2, 0,1]])
print(f'Predicted price of Ford 2.2L: ₹{pred[0]:.2f}')
```

```python
# prepare featrues and target
X2 = df2[['brand', 'engine_size']]
y2 = df2['price']

# OneHotEncoder with ColumnTransformer
ct=ColumnTransformer([
    ('brand_oheeeee', OneHotEncoder(drop='first'), ['brand'])                  # OneHotEncoder similar to get_dummies
], remainder='passthrough')

X2_encoded = ct.fit_transform(X2)           #never pass y
X2_encoded[:5]
```

```python
model2 = LinearRegression()
model2.fit(X2_encoded, y2)
```

```python
print('Model coefficeints:', model2.coef_)
print('Model intercept:', model2.intercept_)
print('R2 score:', model2.score(X2_encoded, y2))
```

```python
pred2 = model2.predict([[1,0,3.0]])           #[brand_bmw, brand_ford, engine_size]
print(f'Predicted price of bmw 3.0L: ₹{pred2[0]:.2f}')
```
