```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Sample Data with missing values
data = {'A': [1, 2, np.nan, 4, 5],
'B': [np.nan, 2, 3, 4, 5],
'C': [1, 2, 3, np.nan, np.nan],
'D': [10, 20, 30, 40, 50]}

df_missing = pd.DataFrame(data)

print("Original DataFrame with missing values:")
print(df_missing)

```

```python
print(df_missing.isnull().sum())
```

```python
plt.figure(figsize= (8,6))
sns.heatmap(df_missing.isnull(), cbar=True, cmap='viridis')
plt.title('Missing values heatmap')
plt.show()
```

```python
#Imputation with mean/median

df_filled_mean = df_missing.copy()
df_filled_mean['A']=df_filled_mean['A'].fillna(df_filled_mean['A'].mean())

print("dataframe after mean imputaion in column 'A':")
print(df_filled_mean)
```

```python
print(df_filled_mean['A'].mean())
```

```python
#Imputing in column B
df_filled_mean['B']=df_filled_mean['B'].fillna(df_filled_mean['B'].mean())

print("dataframe after mean imputaion in column 'B':")
print(df_filled_mean)
```

```python
print(df_filled_mean['B'].mean())
```

```python
# Imputing in col C
df_filled_mean['C']=df_filled_mean['C'].fillna(df_filled_mean['C'].mean())

print("dataframe after mean imputaion in column 'C':")
print(df_filled_mean)
```

```python
print(df_filled_mean['C'].mean())
```

```python

df_filled_mean['D']=df_filled_mean['D'].fillna(df_filled_mean['D'].mean())

print("dataframe after mean imputaion in column 'D':")
print(df_filled_mean)
```

```python
print(df_filled_mean['D'].mean())
```

```python


# Sample Sales Data
sales_data = {'Product_ID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
              'Sales_Amount': [150, 200, 120, 300, 180, 250, 100, 400, 220, 170],
              'Category': ['Electronics', 'Clothing', 'Electronics', 'Home Goods', 'Clothing',
                           'Home Goods', 'Electronics', 'Clothing', 'Electronics', 'Home Goods']}
df_sales = pd.DataFrame(sales_data)

print("Original Sales DataFrame:")
print(df_sales)
```

```python
# Histogram for 'Sales_Amount'
plt.figure(figsize=(8,6))
sns.histplot(df_sales['Sales_Amount'], kde=True, bins=5) # bins standard 5 to 15
plt.title("Distibution of sales amount")
plt.xlabel("Sales Amount")
plt.ylabel("Frequency")
plt.show()

# Kernel Density Estimate curve(Kde)
```

```python
# Box plot for 'sales_amount'
plt.figure(figsize=(8,6))
sns.boxplot(x=df_sales['Sales_Amount'])
plt.title('box plot of sales amount')
plt.xlabel('sales amount')
plt.show()
```

```python
#Bar chart of 'category'
sns.countplot(
    x='Category',
    data=df_sales,
    palette='viridis',
    hue='Category',
    legend=False
)
plt.title('Count of products per category')
plt.xlabel('Category')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()

# Histogram (histplot) → for continuous numerical data, bars represent ranges (bins).

# Countplot (countplot) → for categorical data, bars represent categories directly.
```

```python
# Add 'Advertising_Spend' and 'Region' for Bivariate Analysis
df_sales['Advertising_Spend'] = [50, 60, 45, 80, 55, 70, 40, 90, 65, 50]
df_sales['Region'] = ['North', 'South', 'North', 'East', 'West', 'East', 'North', 'South', 'West', 'East']

print("\nSales DataFrame with new columns:")
print(df_sales)
```

```python
#Scatter plot : sales_amount vs advertising_spend
plt.figure(figsize=(9,7))
sns.scatterplot(x='Advertising_Spend', y='Sales_Amount', data=df_sales, hue='Category', size='Sales_Amount', sizes=(20, 400))
plt.title('Sales Amount vs Advertising Spend')
plt.xlabel('Advertising spend')
plt.ylabel('Sales amount')
plt.show()

#shows positive correlation
```

```python
#Boxplot(numeriacal and categorical)
plt.figure(figsize=(9,7))
sns.boxplot(x='Category', y='Sales_Amount', data=df_sales, palette='pastel', hue='Category',legend=False)
plt.title("Sales amount by product category")
plt.xlabel('Category')
plt.ylabel('Sales_Amount')
plt.show()
```

```python
# cross-tabulation table (also called a contingency table)

cross_tab = pd.crosstab(df_sales['Region'], df_sales['Category'])
print("\n Corss-tabulation of Region and Category")
print(cross_tab)
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(cross_tab, annot=True, fmt='d')
plt.title('Distribution of product categories by region')
plt.xlabel('Category')
plt.ylabel('Region')
plt.show()
```

```python
plt.figure(figsize=(10,8))
sns.pairplot(df_sales, vars=['Sales_Amount', 'Advertising_Spend'], hue='Category',diag_kind='kde')
plt.show()
```

```python
corr_matrix = df_sales[['Sales_Amount', 'Advertising_Spend']].corr()
print(corr_matrix)
```

```python
plt.figure(figsize= (8,6))
sns.heatmap(corr_matrix, annot=True)
plt.title('Correlation heamap')
plt.show()
```

```python
g= sns.FacetGrid(df_sales, col='Region', row='Category', height=3, aspect=1.2)
g.map(sns.histplot, 'Sales_Amount', kde=True, bins=5)
g.set_axis_labels("Sales_amount", "Frequency")
g.set_titles(col_template="{col_name} Region", row_template="{row_name} Category")

plt.tight_layout(rect=[0,0,1,0.98])
plt.show()
```
