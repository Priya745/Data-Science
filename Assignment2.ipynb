### a. Create and Customize Plots using Matplotlib Library:

i. Line Plot

```python
x = np.linspace(-10, 10, 100)
y = x ** 2

plt.figure(figsize=(8,6))
plt.plot(x,y, color='blue')

plt.title("Line Plot of y = x^2", fontsize=16)
plt.xlabel("x values", fontsize=12)
plt.ylabel("y values", fontsize=12)
plt.legend()

plt.savefig("line_plot.png", dpi=300, bbox_inches='tight')
plt.show()

# The graph is a parabola opening upwards.
# Interpretation: The curve shows how small changes near zero produce small outputs, but large inputs produce much larger outputs — 
                 # a visual of exponential-like growth but slower.
```

ii. Bar Plot

```python
data = pd.DataFrame({
    'Category': ['Electronics', 'Clothing', 'Books'],
    'Quantity': [25, 40, 15]
})

print(data)
```

```python


plt.figure(figsize=(8,6))

# Plot each bar separately with labels
plt.bar(data['Category'], data['Quantity'], 
        color=['blue','green','orange'], edgecolor='black')
# Customize
plt.title("Category-wise Quantities")
plt.xlabel("Category")
plt.ylabel("Quantity")
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Add legend
plt.legend(title="Categories")

plt.savefig("bar_chart.png", dpi=300, bbox_inches='tight')
plt.show()
```

iii. Histogram

```python
data = np.random.normal(loc=0, scale=1, size=1000)
# loc = mean (0), scale = standard deviation (1), size = number of samples (1000)

plt.figure(figsize=(8,6))
plt.hist(data, bins=30, color='skyblue')

plt.title("Histogram")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.savefig("histogram.png", dpi=300, bbox_inches='tight')

plt.show()
```

### b. Load, Visualize Data, and Customize Visualizations using Seaborn Library:

i. Load iris dataset from Seaborn

```python
iris = sns.load_dataset("iris")
print(iris.head())
```

ii. Pairplot to visualize the relationships between different features in the dataset.

```python
sns.pairplot(iris, hue="species", palette="husl", diag_kind="kde")

# hue parameter is used to color the data points based on a categorical variable.

plt.suptitle("Pairplot of Iris Dataset")
plt.show()
```

iii. Boxplot to show the distribution of petal lengths for different species in the iris dataset.

```python
plt.figure(figsize=(8,6))
sns.boxplot(data=iris, x="species",y="petal_length",hue="species", palette="Set2")

plt.title("Distribution of Petal Lengths by Species")
plt.xlabel("Species")
plt.ylabel("Petal length (cm)")
plt.show()
```

iv. Regression plot to show the relatioship between sepal_length and sepal_width.

```python
# regplot → single regression line for all data
sns.regplot(x="sepal_length", y="sepal_width", data=iris, scatter_kws={"color":"blue","alpha":0.6},
            line_kws={"color": "red"})

plt.title("Relationship between Sepal length and Sepal width")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.show()
```

```python
#lmplot → regression lines separated by species (better for categorical comparison)
sns.lmplot(x="sepal_length", y="sepal_width", hue="species", data=iris,
           markers=["o", "s", "D"], palette="husl",
           height=5, aspect=1.2)

plt.title("Sepal Length vs Sepal Width by Species")
plt.show()
```

### c. Implement and evaluate a Logistic Regression classifier using Scikit-learn Library:

i. 'iris' dataset from Scikit-learn

```python
from sklearn.datasets import load_iris

iris = load_iris()

```

ii. Split dataset into training and testing sets.

```python
from sklearn.linear_model import LogisticRegression

# Convert to dataframe
X = pd.DataFrame(iris.data, columns=iris.feature_names)
y = pd.Series(iris.target, name="species")

print(X.head())
print(y.head())

X_train, X_test, y_train, y_test = train_test_split(

    X, y, test_size=0.3, random_state=42, stratify=y
)
```

iii. Train a Logistic Regression model on the training set.

```python
#Initialize logistic regression model
log_reg = LogisticRegression(max_iter=200, random_state=42)

log_reg.fit(X_train, y_train)
```

iv. Evaluate the model on the testing set and print the accuracy.

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

y_pred = log_reg.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

### d. Implement and evaluate a Linear Regression model using Scikit-learn library

i. Use 'boston' housing dataset from scikit-learn

```python
from sklearn.datasets import fetch_california_housing

california = fetch_california_housing()

X = pd.DataFrame(california.data, columns=california.feature_names)
y = pd.Series(california.target, name="HouseValue")

print(X.head())
print(y.head())

```

ii. Split the dataset

```python
from sklearn.model_selection import train_test_split

# Assuming you already have X (features) and y (target)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

print("Training set size:", X_train.shape)
print("Testing set size:", X_test.shape)
```

iii. Train a Linear Regression model on training set

```python
from sklearn.linear_model import LinearRegression

# Initialize the model
lin_reg = LinearRegression()

# Train on training set
lin_reg.fit(X_train, y_train)
```

iv. Evaluate the model on training set.

```python
from sklearn.metrics import mean_squared_error, r2_score

# Predictions on training set
y_train_pred = lin_reg.predict(X_train)

# Evaluation metrics
mse_train = mean_squared_error(y_train, y_train_pred)
r2_train = r2_score(y_train, y_train_pred)

print("Training Set Mean Squared Error:", mse_train)
print("Training Set R² Score:", r2_train)
```
